```python
import pandas as pd
from pathlib import Path

def merge_parquet_folder(folder: str) -> pd.DataFrame:
    folder = Path(folder)
    parquet_files = list(folder.glob("*.parquet"))

    if not parquet_files:
        raise FileNotFoundError(f"No parquet files found in {folder}")

    dfs = [pd.read_parquet(f) for f in parquet_files]
    merged = pd.concat(dfs, ignore_index=True)

    return merged

# Example usage
df = merge_parquet_folder("/home/jovyan/nuclei-foundational-model/results/slides")
df.to_parquet("/home/jovyan/nuclei-foundational-model/results/slides.parquet")
print(df.shape, df.head())
```
